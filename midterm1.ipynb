```python
import math
import pandas as pd

# Global settings and constants
Z = 1.96
LANGS = ["EN", "RU", "KK"]
CANDIDATES = [("A", 3), ("A", 5), ("B", 3), ("B", 5)]
RANKS_CSV = "my_data/retrieval_ranks.csv"  # Path to retrieval ranks CSV file in Colab
DOCS_CSV = "my_data/doc_results.csv"      # Path to document results CSV file in Colab

def wilson(hits, n, z=Z):
    """Calculate the 95% Wilson score interval for a proportion."""
    p = hits / n
    denom = 1 + z * z / n
    centre = (p + z * z / (2 * n)) / denom
    half = z * math.sqrt(p * (1 - p) / n + z * z / (4 * n * n)) / denom
    return max(0.0, centre - half), min(1.0, centre + half)

def fmt_ci(hits, n):
    """Format proportion and Wilson confidence interval as a string."""
    lo, hi = wilson(hits, n)
    return f"{hits}/{n} = {hits / n:.3f} [{lo:.3f}, {hi:.3f}]"

def mrr10(ranks):
    """Calculate Mean Reciprocal Rank at k=10."""
    return sum(1 / r for r in ranks if r <= 10) / len(ranks)
```

```python
P = dict(
    share={"EN": 41, "RU": 43, "KK": 16},            # Traffic share percentage
    q_tok={"EN": 36, "RU": 72, "KK": 132},          # Question tokens per language
    a_tok={"EN": 133, "RU": 266, "KK": 488},        # Answer tokens per language
    chunk={"A": {"EN": 269, "RU": 538, "KK": 987},
           "B": {"EN": 67, "RU": 134, "KK": 246}},   # Chunk tokens per config and language
    system=247, rpd=12500, p_in=2.5, p_out=15, budget=32600,
)
Q = dict(pages_day=5000, pii=16, w=1240, h=1754, out_tok=367,
         p_in=3, p_out=15, srv_month=450, srv_cap=8000)
```

```python
def annual_cost(cfg, k):
    """Calculate the annual cost of model calls for a given chunking config and k."""
    per_request = 0.0
    for lang in LANGS:
        tokens_in = P["system"] + P["q_tok"][lang] + k * P["chunk"][cfg][lang]
        cost = (tokens_in * P["p_in"] + P["a_tok"][lang] * P["p_out"]) / 1e6
        per_request += P["share"][lang] / 100 * cost
    return per_request * P["rpd"] * 365
```

```python
ranks = pd.read_csv(RANKS_CSV)
N = len(ranks)

print("=== 1a: Recall@k and MRR@10 ===")
rows = []
for cfg in "AB":
    col = ranks[f"rank_{cfg}"]
    rows.append({"config": cfg,
                 **{f"recall@{k}": round((col <= k).sum() / N, 3) for k in (1, 3, 5)},
                 "MRR@10": round(mrr10(col), 4)})
print(pd.DataFrame(rows).to_string(index=False))

print("\n=== 1b + 1c: Intervals and Annual Cost ===")
cands = []
for cfg, k in CANDIDATES:
    hits = int((ranks[f"rank_{cfg}"] <= k).sum())
    lo, hi = wilson(hits, N)
    cost = annual_cost(cfg, k)
    cands.append(dict(cfg=cfg, k=k, hits=hits, lo=lo, hi=hi, cost=cost,
                      ok=cost <= P["budget"]))
table = pd.DataFrame(cands)
table["recall"] = (table.hits / N).round(3)
table["CI95"] = table.apply(lambda r: f"[{r.lo:.3f}, {r.hi:.3f}]", axis=1)
table["cost_$"] = table.cost.round(0).astype(int)
table["in_budget"] = table.ok
print(table[["cfg", "k", "hits", "recall", "CI95", "cost_$", "in_budget"]].to_string(index=False))
print(f"Budget Limit: ${P['budget']:,}")

print("\n=== 1d: Decision & Selection ===")
affordable = [c for c in cands if c["ok"]]
best = max(affordable, key=lambda c: (c["hits"], -c["cost"]))
pool = [c for c in affordable if c["lo"] <= best["hi"] and best["lo"] <= c["hi"]]
choice = min(pool, key=lambda c: c["cost"])
print(f"Best recall within budget: ({best['cfg']}, {best['k']})")
print("Statistically indistinguishable from best (overlapping intervals):",
      [(c["cfg"], c["k"]) for c in pool])
print(f"Selected candidate (cheapest of the pool): ({choice['cfg']}, {choice['k']}), "
      f"${choice['cost']:,.0f}, budget surplus ${P['budget'] - choice['cost']:,.0f}")

print(f"\nPer-language recall for selected candidate ({choice['cfg']}, {choice['k']}):")
col = f"rank_{choice['cfg']}"
for lang in LANGS:
    sub = ranks[ranks.lang == lang]
    print(f"  {lang}: {fmt_ci(int((sub[col] <= choice['k']).sum()), len(sub))}")
```

```python
docs = pd.read_csv(DOCS_CSV)
docs["acc"] = docs.n_correct / docs.n_docs

print("\n=== 2a: Pipeline Accuracy & Cost ===")
img_tokens = math.ceil(Q["w"] / 28) * math.ceil(Q["h"] / 28)
cost_pipe = {
    "H": Q["pages_day"] * 365 * (img_tokens * Q["p_in"] + Q["out_tok"] * Q["p_out"]) / 1e6,
    "L": math.ceil(Q["pages_day"] / Q["srv_cap"]) * Q["srv_month"] * 12,
}
for pipe in "HL":
    s = docs[docs.pipeline == pipe]
    weakest = s.loc[s.acc.idxmin()]
    print(f"Pipeline {pipe}: overall {fmt_ci(int(s.n_correct.sum()), int(s.n_docs.sum()))}")
    print(f"   weakest cell ({weakest.lang}, {weakest.condition}): "
          f"{fmt_ci(int(weakest.n_correct), int(weakest.n_docs))}")
    print(f"   annual cost: ${cost_pipe[pipe]:,.0f}")
    for lang in LANGS:
        t = s[s.lang == lang]
        print(f"   {lang} (all conditions): {fmt_ci(int(t.n_correct.sum()), int(t.n_docs.sum()))}")

print(f"\nImage tokens per page (Pipeline H): {img_tokens}")
print(f"Pages with PII leaving the country annually under H: "
      f"{Q['pages_day'] * 365 * Q['pii'] / 100:,.0f} pages/year")
print("\nAccuracy pivot table by cell:")
print(docs.pivot_table(index=["lang", "condition"], columns="pipeline", values="acc").round(3))
```
